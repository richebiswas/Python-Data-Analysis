```python
import matplotlib .pyplot as plt
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("pokemon.csv")
```

```python
df
```

```python
type_count=df["Type1"].value_counts(ascending=True)
```

```python
plt.barh(type_count.index,type_count.values,color="pink",
        edgecolor="black",

         )
plt.title("No of pokemon by primary type")
plt.xlabel("Count")
plt.ylabel("Type")

plt.tight_layout()
plt.show()
```
