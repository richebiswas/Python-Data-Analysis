```python
import matplotlib .pyplot as plt
import numpy as np
```

```python
scores=np.random.normal(loc=80,scale=20,size=100) #100 test scores
#loc is the median/mid of our data 
#scale=score is high far away the scores are going to deviate from the centre
#higher the no ,graeter the deviation

scores=np.clip(scores,0,100)
```

```python
plt.hist(scores,bins=10,
         color="lightgreen",
         edgecolor="black",
         )

plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("No of Students")

plt.show()
```
