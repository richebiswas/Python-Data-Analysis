```python
import matplotlib .pyplot as plt
import numpy as np
```

```python
x=np.array([2023,2024,2025,2026])
y1=np.array([15,55,90,190])
y2=np.array([45,55,82,200])
```

# PLOT CUSTOMIZATION

```python


line_style=dict( marker=".",markersize=20,
        markerfacecolor="cyan",markeredgecolor="blue",
        linestyle="solid",linewidth=2)

#unpack the dictionary with **
  
plt.plot(x,y1,color="green",**line_style)
plt.plot(x,y2,color="red",**line_style)
#marker= . o v * 
#linestyle=solid,dashed dotted none
```

# LABELS

```python
plt.plot(x,y1,color="green",**line_style)
plt.plot(x,y2,color="red",**line_style)
plt.title("Class size",fontsize=20,
          family="Arial" ,
          fontweight="bold",
          color="blue")
plt.xlabel("Year",fontsize=10,
          family="Arial" ,
          fontweight="bold",
          color="maroon")
plt.ylabel("Students",fontsize=10,
          family="Arial" ,
          fontweight="bold",
          color="maroon")
plt.tick_params(axis="both",
            colors="blue")
```
