```python
import matplotlib .pyplot as plt
import numpy as np
```

```python
categories=["Grains","Fruits&Vegtables","Proteins","Dairy","Sweets"]
values=np.array([300,275,200,150,100])
colors=["yellow","green","red","pink","purple"]
```

```python
plt.pie(values,labels=categories,
        autopct="%1.1f%%", #automatic percentage
        colors=colors,
        explode=[0.1,0,0,0,0],
        shadow=True,
        startangle=150)
plt.title("Food consumed ")
plt.show()
```
