```python
import numpy as np
import pandas as pd
```

Creating data frame


```python
data = {

   "Name": ["Rahul", "Priya", "Arjun", "Sneha", "Rohan"],

   "Age": [22, 25, 21, 28, 24],

   "City": ["Kolkata", "Mumbai", "Delhi", "Bangalore", "Pune"],

   "Salary": [35000, 50000, 30000, 65000, 45000]

}
```

```python
columns = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(data,columns =columns)
df2
```

Selection and Indexing of Columns

```python
df2[["Name","City"]]
```

Creating a new column


```python
df2["Designation"] = ["Doctor","Eng.","Professor","Developer","Plumber"]
```

```python
df2
```

Removing Columns


```python
df2.drop("Designation",axis=1)  #axis 1 is column and axis 0 is row by deafult pandas take axis=0 if not explicitly mentioned
```

```python
df2.drop("Designation",axis=1,inplace=True) #inplace means drop from original data base
```
