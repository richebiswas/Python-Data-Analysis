# grid  


helps make plots easier to read by adding refernce line

```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])

plt.plot(x,y)
plt.grid(axis="y",
         linewidth=2,
         color="lightgray",
         linestyle="dotted"
         )
plt.show()
```
